# <a name="08intro">PMF vs PDF</a>

---

Since **discrete** random variables have countable individual outcomes, we can come up with exact probabilities for each outcome. We call the distribution of these probabilities the "Probability MASS Function", or PMF. We use "mass" here because each outcome has a "chunk" of the total probability.

Continuous random variables don't have the same neat, easily countable outcomes - the outcomes can be any value within a range. We call the distribution of these probabilities the "Probability DENSITY Function", or PDF. We use "density" here because these probabilities are spread smoothly over a continuum.

| **Characteristic** | **PMF**                                   | **PDF**                                                                                |
| ------------------ | ----------------------------------------- | -------------------------------------------------------------------------------------- |
| **Type**           | Discrete variables                        | Continuous variables                                                                   |
| **Purpose**        | Probability of specific outcomes          | Probability over an interval                                                           |
| **Sum/Integral**   | Sum of probabilities = 1                  | Integral of density = 1                                                                |
| **Units**          | Unitless probabilities                    | Density (inverse of variable's unit)                                                   |


**The main difference here is that with continuous variables, we have to calculate the probability *over an interval* instead of at a point.**

---

See the example below on how to graph the PDF of the continuous function

# <a name="06prop-cont">Mean, Median, and Variance of Continuous Random Variables</a>

---

Let $f(x)$ and $F(x)$ denote the probability density function and
cumulative distribution function for a continuous random variable $X$.

-   The <font color="dodgerblue">**expected value**</font> or <font color="dodgerblue">**mean**</font> is

$$E(X) = \mu_X = \int_{-\infty}^{\infty} x \cdot f(x) \, dx.$$

-   The <font color="dodgerblue">**variance**</font> is

$$\mbox{Var}(X) = E\big[ (X-\mu_X)^2 \big] = E(X^2) - \mu_X^2.$$

-   The <font color="dodgerblue">**median**</font> is the value $x$ such that $P(X < x) = 0.5$. Thus, to find the median we solve the equations below for $x$,

$$\int_{-\infty}^x f(t) \, dt = 0.5 \qquad \mbox{or equivalently} \qquad F(x) =0.5.$$

# <a name="08unif">Continuous Uniform Distribution</a>

---


Similar to its discrete counterpart, $X$ follows a <font color="dodgerblue">**continuous uniform
distribution**</font> when $X$ is equally likely to equal any value
on the closed, continuous interval $\lbrack a , b \rbrack$. If a
continuous random variable is uniformly distributed on the interval
$\lbrack a , b \rbrack$, the pdf is

$$f(x; a,b) = \left\{ \begin{array}{ll} \dfrac{1}{b-a} & a \leq x \leq b\\ 0 & \mbox{otherwise} \end{array} \right. .$$

The corresponding cdf is

$$F(x) = \left\{ \begin{array}{ll}
0, &  x<a \\
\frac{x-a}{b-a}, &  a \leq x \leq b \\
1,  &  x>b
\end{array} \right. .$$

-   The expected value is $E(X) = \dfrac{a+b}{2}$

-   The variance is $\mbox{Var}(X) = \dfrac{(b-a)^2}{12}$

When working with uniform distributions, it is typically easier to calculate probabilities “by hand” without the need for technology. However, R does have functions to help!

-   Use `dunif(x, a, b)` to find the height of the pdf function, which
    is $\frac{1}{b-a}$.
    -   There is really no need to use the function since the pdf has a
        constant height.
    -   <font color="tomato">**DO NOT use `dunif()` to compute probabilities.**</font>
-   Use `punif(x, a, b)` to calculate $P(X \leq x)=\int_a^x \frac{1}{b-a} \, dt = \frac{x-a}{b-a}$.
    -   **The probability for a uniform distribution can be found by computing the area of a rectangle which may be easier!**
-   Use `runif(n, a, b)` to generate a random sample of size $n$ from population $X \sim \mbox{Unif}(a, b)$.
-   Use `qunif(q, a, b)` to find the q<sup>th</sup> percentile.

<br>
The code below shows the PDF and CDF of an example continuous uniform distribution:

$ f_X(x) = \left\{ \begin{array}{ll}
\frac{1}{2}, &  3 \leq x \leq 5 \\
0, &  \mbox{otherwise}
\end{array} \right. .$

```r
x <- seq(2.5, 5.5, length=1000)


f = function(x)
{ (x<3) * (0) + (x >= 3 & x <= 5) * (dunif(x, 3, 5)) + (x > 5) * (0) }

plot(x, f(x), type="l", col = "red", xlab="x value",
  ylab="Density", main="pdf Uniform on [3,5]",
  xlim = c(2.5, 5.5), ylim = c(0, 0.75))

g = function(x)
{
  (x < 3) * (0) + (x >= 3 & x <= 5) * (punif(x, 3, 5)) + (x > 5) * (1)
}

plot(x, g(x), type="l", col = "blue", xlab="x value",
  ylab="Density", main="cdf Uniform on [3,5]",
  xlim = c(2.5, 5.5))
```

---

There are many situations where data tends to be distributed
symmetrically around a central value with no bias to the left or right
such as the location of beads on a [Galton
board](https://en.wikipedia.org/wiki/Galton_board) or heights of people.
For example, if we randomly select an adult from the population:

-   A person is most likely to be about average height.
-   Otherwise, the person selected is equally likely to be either shorter or taller than average.
-   The person is very unlikely to have an extreme height (tall or short). The more extreme, the less likely.
-   **Therefore the distribution of heights is approximately symmetric and bell shaped.**

<font color="dodgerblue">**Normal distributions**</font> arise in
many settings: heights of people, size of items produced by machines,
and most importantly in statistics, data sets resulting from many
independent random events. The shape of a normal distribution is
determined by two parameters:

-   The <font color="dodgerblue">**mean**</font>, $\color{dodgerblue}{\mu}$, is the center of the distribution.
-   The <font color="dodgerblue">**standard deviation**</font> $\color{dodgerblue}{\sigma}$, measures the spread or variability of the data.
-   If $X$ is normally distributed with mean $\mu$ and standard deviation $\sigma$, then we write $\color{dodgerblue}{\mathbf{X \sim N(\mu, \sigma)}}$.



## <a name="08append-normal">Properties of Normal Distributions</a>

---

If $X$ is <font color="dodgerblue">**normally distributed**</font>
with mean $\mu$ and standard deviation $\sigma$, we write
$\color{dodgerblue}{X \sim N( \mu ,\sigma)}$. The formula for the
probability density is called the Gaussian and is given below.

$$f(x; \mu, \sigma) = \dfrac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2}$$

-   The expected value is $E(X) = \mu$.
-   The variance is $\mbox{Var} (X) = \sigma^2$.
-   Use `dnorm(x, mu, sigma)` to calculate $f(x)$.
    -   <font color="tomato">Recall for continuous distributions,
        $f(x)$ gives the height of the pdf.</font>
    -   <font color="tomato">Probabilities are areas under $f(x)$
        (not the heights).</font>
    -   <font color="tomato">**DO NOT use `dnorm()` to compute
        probabilities.**</font>
-   <font color="mediumseagreen">**Be sure to use `pnorm(x, mu, sigma)` to calculate $P(X \leq x)$, the area to the left of $X=x$.**</font>
    -   <font color="mediumseagreen">**One last warning, be sure to use `pnorm()` to compute probabilities not `dnorm()`.**</font>
-   Use `rnorm(n, mu, sigma)` to generate a random sample of size $n$ from population $X \sim N(mu, sigma)$.
-   Use `qnorm(q, mu, sigma)` to find the q<sup>th</sup> percentile.
-   We have the **[68%-95%-99.5% Empirical Rule](#08empirical)**</font>.
-   We can standardize the normal distribution using **[z-scores](#08z-score)**:
    $$z = \frac{x - \mu}{\sigma}$$



```r
x <- seq(15, 65, length=1000)

plot(x, dnorm(x, 40, 5), type="l", col = "red", xlab="x value",
  ylab="Density", main="pdf for N(40,5)")

plot(x, pnorm(x, 40, 5), type="l", col = "blue", xlab="x value",
  ylab="Density", main="cdf for N(40,5)")
```

## <a name="08append-normal">How the standard deviation $\sigma$ affects the distribution shape</a>

---

Since the standard deviation $\sigma$ is a measure of how spread out the values in a distribution are, it can affect the shape of the distribution.

- If $\sigma$ is *small*, it means that the values are close to the mean, making the distribution **tall and skinny**.

- If $\sigma$ is *large*, it means that the values are far away from the mean, making the distribution **shorter and wider**.

```r
# run code cell to compare distributions
# with different standard deviations
# nothing to edit in this code cell

x <- seq(-10, 10, length=1000)
stan <- dnorm(x)

std <- c(1, 2, 4)
colors <- c("black", "blue", "red")
labels <- c("sigma=1", "sigma=2", "sigma=4")

plot(x, stan, type="l", lty=2, xlab="x value",
  ylab="Density", main="Comparison of St Dev")

for (i in 1:3){
  lines(x, dnorm(x, 0, std[i]), lwd=2, col=colors[i])
}

legend("topright", inset=.05, title="Distributions",
  labels, lwd=2, lty=1, col=colors)
```

## <a name="08append-normal">How the mean $\mu$ affects the distribution shape</a>

---

Since normal distributions center around the mean, its value can shift the distribution along the x axis.

- If $\mu$ is 0 (or close enough to it), the distribution will be **symmetrical**.

- If $\mu$ is *positive*, the distribution will shift to the **right**.

- If $\mu$ is *negative*, the distribution will shift to the **left**.

*Note: the mean does not affect the shape of the distribution, just its location on the horizontal axis.*

```r
# run code cell to compare distributions
# with different means
# nothing to edit in this code cell

x <- seq(-10, 10, length=100)
stan <- dnorm(x)
mu <- c(0, 5, -5)
colors <- c("black", "blue", "red")
labels <- c("mu=0", "mu=5", "mu=-5")

plot(x, stan, type="l", lty=2, xlab="x value",
  ylab="Density", main="Comparison of Means")

for (i in 1:3){
  lines(x, dnorm(x, mu[i], 1), lwd=2, col=colors[i])
}

legend("topright", inset=.05, title="Distributions",
  labels, lwd=2, lty=1, col=colors)
```

# <a name="08empirical">The 68%-95%-99.7% Empirical Rule</a>

---

The <font color="dodgerblue">**empirical rule**</font> for normal
distributions states that approximately:

-   68% of all values fall within one standard deviation (both above and
    below) from the mean.
-   95% of all values fall within two standard deviations of the mean.
-   99.7% of all values fall within three standard deviations of the
    mean.

# <a name="08z-score">$z$-Scores and the Standard Normal Distribution</a>

---

The number of standard deviations $x$ is from the mean
$\mu$ is called the <font color="dodgerblue">**z-score**</font>
for the value $x$.

- We compute the signed distance of $x$ from the mean, $\mu$.
- Then we describe the distance in terms of standard deviations (as opposed to in units of $x$).
- The formula for the $z$-score of $x$ is therefore

$$\color{dodgerblue}{\large z = \frac{x-\mu}{\sigma}}.$$

For example, the $z$-score of an adult with a total cholesterol level of
140 mg/dL is $z=\frac{140-200}{40} = -1.5$. This means a total cholesterol level of 140 mg/dL is 1.5 standard deviations below the mean level.

<figure>
<img
src="https://upload.wikimedia.org/wikipedia/commons/a/a4/08fig-standardize.png"
data-fig-alt="Standardizing a normal distribution"
alt="Image Credit: Adam Spiegler, CC BY-SA 4.0, via Wikimedia Commons"  width="80%"/>
<figcaption aria-hidden="true">Image Credit: Adam Spiegler, <a
href="https://creativecommons.org/licenses/by-sa/4.0">CC BY-SA 4.0</a>,
via Wikimedia Commons</figcaption>
</figure>

- Comparing areas in the figure above, we see that $P(X<140) = P(Z < -1.5) $.
- <font color="dodgerblue">**Thus, using either the original distribution or the distribution of $z$-scores gives the same result.**</font>


When we compute the $z$-score, we are <font color="dodgerblue">**“standardizing”**</font> the normally
distributed data. We describe values of random variable $X$ in terms of
how many standard deviations each value $x$ is from the mean $\mu$. The
resulting distribution of $z$-scores is called the <font color="dodgerblue">**standard normal distribution**</font> that
has mean $0$ and standard deviation $1$.

# <a name="08normal-pdf">The Probability Density Function for Normal Distributions</a>

---

The probability density function for a normal distribution
$X \sim N(\mu,\sigma)$ is called a <font color="dodgerblue">**Gaussian function**</font> whose formula is

$$\color{dodgerblue}{\large f(x; \mu, \sigma) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2}}.$$

<font color="mediumseagreen">*Note: The formula for a Gaussian depends on both irrational numbers $\pi$ and $e$. How surprising and cool is that!*</font>



# <a name="08normal-area">Calculating Areas Under Normal Distributions</a>

---

To find the probability that an adult woman in Israel has a BMI less
than $18.5$, we need to calculate the area to the left of $x=18.5$ under
the normal curve with formula $f(x; 26.5, 4.5)$. We can try to evaluate

$$P(X < 18.5) = \int_{0}^{18.5} \frac{1}{4.5\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-26.5}{4.5} \right)^2} \, dx$$

YIKES, good luck with that! We cannot find a formula for the

## <a name="08normal-R">Using R to Compute Areas</a>

---

Standard normal distribution tables were very useful prior to spread
statistical software and technology. If we are working in R, we do not
need to use tables to get approximations. Instead, we can use R
functions to compute these areas more accurately.

Similar to binomial, uniform, and Poisson distributions, R has built
in functions to perform calculations with normal distributions. In
particular, the function `pnorm()` calculates values of the cumulative
distribution function for a normal distribution:

-   `pnorm(x, mean, sd)`$=P(X<x)$ gives the area to the left of $x$ under $N(\mu,\sigma)$.
-   `pnorm(x, mean,  sd, lower.tail=FALSE)`$=P(X>x)$ gives the area to the right of $x$ under $N(\mu,\sigma)$. (Note that you can also calculate this by finding $1-P(X<x)$ = 1 - `pnorm(x, mean,  sd)` )

Or you can find the $z$-score and use the standard normal distribution
in R as well:

-   `pnorm(z, 0, 1)`$=P(Z<z)$ gives the area to the left of $x$ under $N(0,1)$.
-   `pnorm(z, 0,  1, lower.tail=FALSE)`$=P(Z>z)$ gives the area to the right of $x$ under $N(0,1)$.

<font color="mediumseagreen">**Remember: Unlike discrete distributions: Do not** use the function `dnorm(x, mean, sd)` to compute **probabilities**.

 `dnorm()` returns the height of the
probability density function, for continuous distributions probability is always that area under the distribution so you would always use `pnorm()`.</font>
