# <a name="07append-unif">The Discrete Uniform Distribution</a>

---

In a discrete uniform distribution, there are $k$ distinct outcomes each with an equal likelihood of occurring.

$$f(x; k) = \left\{ \begin{array}{ll} \dfrac{1}{k} \ \ & \mbox{for } x = 1, 2, \ldots,  k\\ 0 \ \ & \mbox{otherwise} \end{array} \right.$$

-   The expected value is $\displaystyle E(X) = \mu_X = \frac{k+1}{2}$.
-   The variance is $\displaystyle \mbox{Var} (X) = \sigma_X^2 = \frac{k^2-1}{12}$.

Examples: a fair dice, a fair coin...

*Note functions such as `dunif()`, `punif()`, `runif()`, and `qunif()` are for the **continuous** (not discrete) uniform distribution. It is typically easier to calculate probabilities directly rather than use technology.*

<br>

This is the PMF and CDF for a fair, 6-sided die:


```r
x.min <- 1
x.max <- 6
x <- x.min:x.max  # vector of integers  x.min,  ..., x.max
p.x <- rep(1/(x.max - x.min+1), length(x) )
c.x <- p.x * (x)

par(mfrow = c(1, 2))

plot(x, p.x,
     type='p',  # point plot
     main='PMF',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels

plot(x, c.x,
     type='s',  # stairs plot
     main='CDF',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels
```

# <a name="07append-pois">The Poisson Distribution</a>

---
### Intuitively...

The Poisson distribution is used to count random events.

It's helpful when we want to answer questions like:
- How many patients might come into the emergency department in the next hour?
- How many cars may cross an intersection tomorrow?
- How many computer crashes could happen in a company next year?

You don't know exactly when each one will happen, but you know the <font color="dodgerblue">**average rate**</font>, which we will call $\lambda$.

So, if an event is *random, spread out over time, and has a predictable average rate*, then it follows a Poisson pattern.

<font color="dodgerblue">**The Poisson distribution tells you how likely you are to observe a certain number of occurences in a fixed interval when you know the average rate.**</font>

Poisson distributions are very useful in biomedical sciences because biology often involves counting rare, random events over time, area, or volume.

*Note that, unlike binomial distributions, there is no fixed number of trials here - just events occurring over a specified range.*

<br>

### Formally...

$X \sim \mbox{Pois}(\lambda)$ is the number of times an event occurs in
a fixed time period, where $\lambda$ denotes the mean number of
occurrences in the given time period.

$$f(x; \lambda) = \left\{ \begin{array}{ll}  e^{-\lambda} \dfrac{\lambda^x}{x!} \ \ & \mbox{for } x =  0, 1, 2, \ldots  \\
0 & \mbox{otherwise} \end{array} \right. .$$

In this equation...
- $e^{-\lambda}$ is the "decay" factor, it makes sure the probabilities of all possible $x$ add up to 1. Basically, as the number of events $x$ gets further away from $\lambda$, the probability will get closer and closer to 0.
- $\lambda^x$ is the "weight" of $x$ events - the more events you want to know about, the less likely it is if $\lambda$ is small, and the more likely it is if $\lambda$ is large.
- dividing by $x!$ adjusts for the fact that order doesn't matter by dividing by all the different ways the same outcome could be "ordered".

Putting it all together, the poisson equation calculates the probability of exactly $x$ events occuring by doing:

$$(probability \,of \,not \,having \,extra \,events) * \frac{(how \,likely \,each \,event \,is \,to \,occur)^x}{number \,of \,ways \,the \,x \,events \,can \, be \,arranged} $$

<br>

The <font color="dodgerblue">**expected value**</font> is $E(X) = \mu_X = \lambda$.

The <font color="dodgerblue">**variance**</font> is $\mbox{Var} (X) = \sigma_X^2 = \lambda$.

## <a name="07append-pois"> R Functions
-   Use `dpois(x, lambda)` to calculate $p(x) = P(X = x)$.
-   Use `ppois(x, lambda)` to calculate $F(x) = P(X \leq x)$.
-   Use `rpois(m, lambda)` to generate a random sample of size $m$ from
    population $X \sim \mbox{Pois}(\lambda)$.
-   Use `qpois(q, lambda)` to find the q<sup>th</sup> percentile of
    $X \sim \mbox{Pois}(\lambda)$.

<br>

Below are the PMF and CDF of a Poisson distribution of $\lambda$ = 5.

```r
x <- 0:15  # creates vector of possible occurrences
lam <- 5  # on average 5 occurrence per time period

par(mfrow = c(1, 2))

plot(x, dpois(x, lam),
     type='p',  # point plot
     main='PMF',
     xlab='x occurences per interval',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels

plot(x, ppois(x, lam),
     type='s',  # stairs plot
     main='CDF',
     xlab='x occurences per interval',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels
```

## <a name="07append-pois">How $\lambda$ affects the PMF</a>

---

Poisson distributions center around $\lambda$ - since that is the average rate the event occurs, it's the most likely outcome.
- Example: if you know that the average number of customers that enter the store between 12 and 1 pm is 20, then the most likely number of customers to walk into the store within that time frame tomorrow is 20.

<br>

<img src="https://github.com/gibsonea/Biostats/blob/main/Images/poisson.png?raw=true" width=400/>

- When $\lambda$ is *small*, the distribution will be **narrow and appear to be skewed right** since most of the probabilities are pulled toward small values of $x$.
    - Imagine $\lambda$ = 1. This means that, most of the time, you'll see 0 or 1 occurrences, and higher occurrences are less and less likely.
- As $\lambda$ *increases*, the distribution will spread out.
    - Imagine $\lambda$ = 4. This means that values near 4 are most likely - 9 or 10 are still possible, but less likely. Still skewed right, but less than before.
- When $\lambda$ is *large*, the distribution spreads out even more and looks flatter. The skew disappears more and gets closer to a symmetrical curve.
    - Imagine $\lambda$ = 20. When events are very frequent, fluctuations above and below $\lambda$ balance out on both sides more often.


## <a name="07append-pois">Example: Patients in the ER</a>

---

Let's assume that, on average, **10 patients come into the ER per hour**.

Our PMF and CDF will look like this:

```r
x <- 0:22  # creates vector of possible occurrences
lam <- 10  # on average 5 occurrence per time period

par(mfrow = c(1, 2))

plot(x, dpois(x, lam),
     type='p',  # point plot
     main='PMF',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels

plot(x, ppois(x, lam),
     type='s',  # stairs plot
     main='CDF',
     cex.lab=1.5, cex.axis=1.5, cex.main=1.5)  # increase font size on labels
```

Using this, we can answer questions like...

- What is the probability that 8 patients will come into the ER next hour?
    - $P(X = 8) = 0.116$
- What is the probability that at most 15 patients will come into the ER next hour?
    - $P(X \leq 15) = 0.951$
- What is the probability that at least two patients will come into the ER next hour?
    - $1 - P(X = 0) = 0.998 $
- ... and so on.


```r
dpois(8, 10)
ppois(15, 10)
1 - dpois(2, 10)
```

### Something to Note

--------------

The graph above is cut off at x = 22 for visual clarity, but in reality...

```r
dpois(0:40, lam)
```

If we extend X to 40, you can see that the probability of x being higher doesn't quite hit 0, just becomes exponentially less likely.

Technically speaking, **even for large $\lambda$, the Poisson distribution can stretch infinitely to the right.**

This does make intuitive sense if we think about it - there is a possiblity that, suddenly, something will happen (a major car accident, maybe) and 40 patients will come into the ER. Way less likely than 20 patients, but still not 0%.

Same with no people coming into a store on Friday at all - even if they normally see an average of 100 people on Fridays. Very unlikely, but not 0%.

## <a name="07latex">Typsetting Arrays in LaTeX</a>

---

-   The `array` environment is started with `\begin{array}`.
-   Next we indicate how many columns and how each column is aligned.
    -   `{ll}` means two columns aligned to the left.
    -   `{lrc}` would be three columns: the first aligned left, then right, then the last column is centered.
    -   We use the `&` symbol to indicate a column break.
    -   We use `\\` to indicate a row break.
-   If we want a big curly brace on the left of the array, enter `\left\{` before beginning the array.
-   We do not want a brace on the other side of the array, so we use `\right.` to close off the left brace without using any matching symbol on the right.
-   We use `\mbox{otherwise}` to type the text *otherwise* inside the equation.
